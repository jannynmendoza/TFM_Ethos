Se construyen las variables que soportan los objetivos 2 y 3 

**Grupos de productos**: cuentas (`ind_cuenta*`), tarjetas/segmento (`ind_cv*`) y productos de inversión/ahorro/protección.
**`tipo_inversion`**: nivel de vinculación patrimonial del cliente (Juvenil → Conservador → Pensionado), construido a partir de `totalpasta` (activo total gestionado), que es la variable que mejor separa los códigos reales de cartera (`tipcart`).
**`segmento_edad`**: Joven / Adulto/ Senior.
**`num_prod_inversion`**: nº de productos de inversión/ahorro que ya tiene el cliente.
**`tiene_inversion`**: variable objetivo binaria para el modelo de jóvenes.
'gastos_tot' llega como texto con formato moneda (" euro  1,278.64", " euro    -  ") -> convertir a numérico

```python
# Función para calcular tipo de inversión según edad
def calcular_tipo_inversion(edad):
    """Calcular tipo de inversión según edad"""
    try:
        edad_val = float(edad)
        if edad_val <= 18:
            return 'Inversión Perfil Juvenil'
        elif edad_val <= 30:
            return 'Inversión Perfil Conservador'
        else:
            return 'Inversión Perfil Pensionado'
    except:
        return 'Inversión Perfil Conservador'  # Valor por defecto

# Calcular tipo de inversión para todos los clientes
df_datasabadell['tipo_inversion'] = df_datasabadell['edad'].apply(calcular_tipo_inversion)

# Mostrar estadísticas de distribución
print("Distribución de tipos de inversión:")
distribucion = df_datasabadell['tipo_inversion'].value_counts()
for tipo, count in distribucion.items():
    porcentaje = (count / len(df_datasabadell)) * 100
    print(f"  - {tipo}: {count} clientes ({porcentaje:.1f}%)")
```

```python
# Función para realizar group by por tipo de cuenta y contar cuántas personas tienen ese tipo de cuenta (sumar cuando el tipo de cuenta sea igual a 1)
def group_by_tipo_cuenta(df_datasabadell, account_columns):
       
    # Verificar que las columnas existan en el DataFrame
    missing_columns = [col for col in account_columns if col not in df_datasabadell.columns]
    if missing_columns:
        print(f"Advertencia: Las siguientes columnas no existen en el DataFrame: {missing_columns}")
        # Remover columnas faltantes de la lista
        account_columns = [col for col in account_columns if col not in missing_columns]
    
    # Crear un DataFrame para almacenar los resultados
    results = []
    
    for account_type in account_columns:
        # Contar cuántas personas tienen esta cuenta (valor = 1)
        count = df_datasabadell[df_datasabadell[account_type] == 1][account_type].count()
        
        # Calcular porcentaje del total
        total_clients = df_datasabadell['id_cliente'].nunique()  # Usar nunique para contar clientes únicos
        percentage = (count / total_clients) * 100 if total_clients > 0 else 0
        
        results.append({
            'Tipo de Cuenta': account_type,
            'Cantidad': count,
            'Porcentaje': f"{percentage:.2f}%"
        })
    
    # Crear DataFrame de resultados
    results_df_datasabadell = pd.DataFrame(results)
    
    # Ordenar por cantidad de personas de mayor a menor
    results_df_datasabadell = results_df_datasabadell.sort_values(by='Cantidad', ascending=False)
    
    return results_df_datasabadell

columnas_cuenta = [
    'ind_cuentaexpansion',
    'ind_cuentaexpansionnegocios',
    'ind_cuentaexpansionpremium',
    'ind_cuentaexperiencia',
    'ind_cuentaprimera',
    'ind_cuentaproyeccion',
    'ind_cvautonomos',
    'ind_cvdirect',
    'ind_cvhabit',
    'ind_cvjove',
    'ind_cvjunior',
    'ind_cvmas',
    'ind_cvprestige',
    'ind_cvsenior',
    'ind_estalinv',
    'ind_estalvi',
  
]
```

```python

# Usar la función con df_datasabadell en lugar de df_tipocuenta
# IMPORTANTE: Primero el DataFrame, luego las columnas
resultados = group_by_tipo_cuenta(df_datasabadell, columnas_cuenta)
print(resultados)
```

kpi_distribucion_productos, Distribución de clientes por nº de productos, este KPI permite Identificar cuántos clientes están mono-producto (riesgo de fuga) vs. muy vinculados 

Definición del KPI:
    numero_productos = total_productos_activos = suma de los 45 indicadores
        ind_* (0/1) de tenencia de producto, por cliente. Se calcula igual
        que en el ETL (staging.sp_cargar_fact_cliente_snapshot / etl_carga_datos.py):
        una simple suma fila a fila, no una fórmula de negocio distinta.

    Para cada valor de numero_productos (0, 1, 2, ...):
        total_clientes   = COUNT(*)
        patrimonio_medio = AVG(total_patrimonio)   -> columna 'totalpasta' del Excel
        margen_medio     = AVG(margen_comercial)   -> columna 'margecomercialut' del Excel



```python
import pandas as pd

# Los mismos 45 indicadores que en etl_carga_datos.py (IND_COLS) y en
# dimensiones.dim_producto (02_dimensiones.sql)
IND_COLS = [
    "ind_cuentaexpansion", "ind_cuentaexpansionnegocios", "ind_cuentaexpansionpremium",
    "ind_cuentaexperiencia", "ind_cuentaprimera", "ind_cuentaproyeccion",
    "ind_cvautonomos", "ind_cvdirect", "ind_cvhabit", "ind_cvjove", "ind_cvjunior",
    "ind_cvmas", "ind_cvprestige", "ind_cvsenior", "ind_estalinv", "ind_estalvi",
    "ind_altrescc", "ind_asvidaestgarcol", "ind_asvidaestulinkc", "ind_bonsestruc",
    "ind_cialp", "ind_credf", "ind_credv", "ind_diposit", "ind_fi", "ind_fialtent",
    "ind_gcfisbp", "ind_gransel", "ind_hipfix", "ind_hipmixt", "ind_hipoteca",
    "ind_hipvar", "ind_pp", "ind_ppassoci", "ind_ppemp", "ind_pprest", "ind_promotor",
    "ind_renttempcol", "ind_renttempind", "ind_rentvehic", "ind_rentvitindiv",
    "ind_restorvrf", "ind_rfixaval", "ind_varispassiu", "ind_vrdafix",
]


def calcular_kpi_distribucion_productos(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()

    # 1. numero_productos: misma suma fila a fila que hace el ETL en SQL
    df["numero_productos"] = df[IND_COLS].sum(axis=1)

    # 2. agregación equivalente al GROUP BY del SQL
    kpi = (
        df.groupby("numero_productos")
        .agg(
            total_clientes=("id_cliente", "count"),
            patrimonio_medio=("totalpasta", "mean"),
            margen_medio=("margecomercialut", "mean"),
           
        )
        .reset_index()
        .sort_values("numero_productos")
        .reset_index(drop=True)
    )
    return kpi


if __name__ == "__main__":
    import sys
    excel_path = sys.argv[1] if len(sys.argv) > 1 else "datasabadell_actualizado_con_segmentos.xlsx"

    from pathlib import Path
    excel_path = Path("datasabadell_actualizado_con_segmentos.xlsx")
    df = pd.read_excel(excel_path, engine='openpyxl')

    
    kpi = calcular_kpi_distribucion_productos(df)

    pd.set_option("display.float_format", lambda v: f"{v:,.2f}")
    print(kpi.to_string(index=False))
```

Este KPI es la base para priorizar en qué segmento de renta enfocar comercialmente — ya sea por volumen (Bajo/Medio-Bajo), por rentabilidad por cliente (Medio-Alto), o por oportunidad de cross-sell no explotado (Alto).
 Permite identificar cómo se comporta la cartera de clientes según su nivel socioeconómico (renta), cruzando ese perfil con patrimonio, rentabilidad y vinculación. En concreto te sirve para:
•	Concentración de la cartera por poder adquisitivo
•	Dónde está la rentabilidad real del banco.
•	Nivel de vinculación (cross-sell) por poder adquisitivo.


```python
#vw_kpi_por_nivel_renta

import pandas as pd

# Mismos 45 indicadores de producto usados en total_productos_activos
# (ver kpi_distribucion_productos.py / 04_staging_etl.sql)
IND_COLS = [
    "ind_cuentaexpansion", "ind_cuentaexpansionnegocios", "ind_cuentaexpansionpremium",
    "ind_cuentaexperiencia", "ind_cuentaprimera", "ind_cuentaproyeccion",
    "ind_cvautonomos", "ind_cvdirect", "ind_cvhabit", "ind_cvjove", "ind_cvjunior",
    "ind_cvmas", "ind_cvprestige", "ind_cvsenior", "ind_estalinv", "ind_estalvi",
    "ind_altrescc", "ind_asvidaestgarcol", "ind_asvidaestulinkc", "ind_bonsestruc",
    "ind_cialp", "ind_credf", "ind_credv", "ind_diposit", "ind_fi", "ind_fialtent",
    "ind_gcfisbp", "ind_gransel", "ind_hipfix", "ind_hipmixt", "ind_hipoteca",
    "ind_hipvar", "ind_pp", "ind_ppassoci", "ind_ppemp", "ind_pprest", "ind_promotor",
    "ind_renttempcol", "ind_renttempind", "ind_rentvehic", "ind_rentvitindiv",
    "ind_restorvrf", "ind_rfixaval", "ind_varispassiu", "ind_vrdafix",
]


MAPA_NIVEL_RENTA = {
    0: "Sin datos",
    1: "Bajo",
    2: "Medio-Bajo",
    3: "Medio",
    4: "Medio-Alto",
    5: "Alto",
}

# Orden lógico para mostrar la tabla (no es alfabético)
ORDEN_NIVEL_RENTA = ["Sin datos", "Bajo", "Medio-Bajo", "Medio", "Medio-Alto", "Alto"]


def calcular_kpi_por_nivel_renta(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()

    # 1. Decodificar nivren -> nivel_renta (igual que el CASE de SQL;
    #    cualquier código fuera de 0-5 o nulo cae en 'Sin datos')
    df["nivel_renta"] = df["nivren"].map(MAPA_NIVEL_RENTA).fillna("Sin datos")

    # 2. total_productos_activos = suma de los 45 indicadores ind_*
    df["total_productos_activos"] = df[IND_COLS].sum(axis=1)

    # 3. Agregar por nivel_renta
    kpi = (
        df.groupby("nivel_renta")
        .agg(
            total_clientes=("id_cliente", "count"),
            patrimonio_medio=("totalpasta", "mean"),
            margen_medio=("margecomercialut", "mean"),
            productos_medios=("total_productos_activos", "mean"),
        )
        .reindex(ORDEN_NIVEL_RENTA)  # mismo orden lógico, aunque no haya datos en algún nivel
        .dropna(how="all")
        .reset_index()
    )
    return kpi


if __name__ == "__main__":
    import sys

    excel_path = sys.argv[1] if len(sys.argv) > 1 else "datasabadell_actualizado_con_segmentos.xlsx"

    from pathlib import Path
    excel_path = Path("datasabadell_actualizado_con_segmentos.xlsx")
    df = pd.read_excel(excel_path, engine='openpyxl')

    kpi = calcular_kpi_por_nivel_renta(df)

    pd.set_option("display.float_format", lambda v: f"{v:,.2f}")
    print(kpi.to_string(index=False))
```
